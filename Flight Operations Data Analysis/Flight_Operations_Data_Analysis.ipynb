```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('Day10_Flight_Operations_Dataset.csv')

# Display the first few rows to understand the structure
display(df.head())
```

```python
print("--- Dataset Information ---")
df.info()

print("\n--- Summary Statistics ---")
display(df.describe(include='all'))

print("\n--- Missing Values Check ---")
print(df.isnull().sum())
```

```python
# 1. Fill missing values in 'Delay_Minutes' with 0 (Assuming NaN means no delay)
df['Delay_Minutes'] = df['Delay_Minutes'].fillna(0)

# 2. Create a new column 'Total_Revenue' (Passengers * Average_Ticket_Price)
df['Total_Revenue'] = df['Passengers'] * df['Average_Ticket_Price']

# 3. Create a 'Route' column by concatenating Origin and Destination
df['Route'] = df['Origin'] + " to " + df['Destination']

# 4. Calculate 'Load_Factor' percentage (Passengers / Seat_Capacity)
df['Load_Factor'] = (df['Passengers'] / df['Seat_Capacity']) * 100

print("Data cleaning and transformations applied. New columns added.")
display(df[['Flight_ID', 'Route', 'Total_Revenue', 'Load_Factor']].head())
```

```python
# Filter: Identify significantly delayed flights (Delay > 30 minutes)
significantly_delayed = df[df['Delay_Minutes'] > 30]
print(f"Total flights delayed by more than 30 minutes: {len(significantly_delayed)}")

# Sort: Top 5 most delayed flights
top_delayed = df.sort_values(by='Delay_Minutes', ascending=False).head(5)
print("\n--- Top 5 Most Delayed Flights ---")
display(top_delayed[['Flight_ID', 'Airline', 'Route', 'Delay_Minutes', 'Weather', 'Flight_Status']])

# Filter & Sort: Cancelled flights sorted by date
cancelled_flights = df[df['Flight_Status'] == 'Cancelled'].sort_values(by='Flight_Date')
print(f"\n--- Cancelled Flights ({len(cancelled_flights)}) ---")
display(cancelled_flights[['Flight_Date', 'Airline', 'Route', 'Weather']])
```

```python
# Group by Airline: Calculate total flights, passengers, revenue, and average satisfaction
airline_performance = df.groupby('Airline').agg(
    Total_Flights=('Flight_ID', 'count'),
    Total_Passengers=('Passengers', 'sum'),
    Average_Load_Factor=('Load_Factor', 'mean'),
    Total_Revenue=('Total_Revenue', 'sum'),
    Avg_Satisfaction=('Passenger_Satisfaction', 'mean')
).sort_values(by='Total_Revenue', ascending=False)

print("\n--- Airline Performance Overview ---")
display(airline_performance)

# Group by Weather: See how weather impacts average delays
weather_delays = df.groupby('Weather')['Delay_Minutes'].mean().reset_index().sort_values(by='Delay_Minutes', ascending=False)

print("\n--- Impact of Weather on Delays ---")
display(weather_delays)

# Group by Route: Most popular routes by flight frequency
popular_routes = df['Route'].value_counts().reset_index().head(5)
popular_routes.columns = ['Route', 'Flight_Count']
print("\n--- Top 5 Most Popular Routes ---")
display(popular_routes)
```

# Key Observations & Findings

*   Vistara leads the market in total flights (37) and total revenue (₹28,689,467), followed closely by IndiGo which carried the most passengers overall (4,158).

*   Interestingly, Air India records the highest average passenger satisfaction score (3.75 out of 5), despite operating fewer flights (24) compared to Vistara and IndiGo.

*   The vast majority of flights operate normally, with 114 out of 180 flights documented as "On Time". Only 7 flights in the entire dataset were "Cancelled".

*   As expected, weather plays a role in flight punctuality. Flights operating in Rain experienced the highest average delay times (21.2 minutes), followed by Storms (19.4 minutes). Surprisingly, Fog had the lowest average delay profile (15.3 minutes).

*   The highest trafficked routes clearly connect major metropolitan hubs. Delhi to Chennai (17 flights) and Mumbai to Bengaluru (16 flights) are the top two most frequently flown routes in this dataset.

*   IndiGo operates with the highest average Load Factor (approx. 66.78%), meaning they are consistently filling more of their aircraft's seat capacity compared to competitors like Akasa Air, which averages the lowest load factor (55.21%).

*   The dataset contains 180 entries, with a high degree of completeness. Only the Delay_Minutes column had missing values (7 records), which were logically filled with 0s indicating no delay occurred.
